```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('academic_performance.csv')
df
```

```python
# Display dataset
print("First 5 rows of dataset:")
print(df.head())
```

```python
# Check dimensions of dataset
print("\nShape of dataset (rows, columns):")
print(df.shape)

# Check data types
print("\nData types of each column:")
print(df.dtypes)

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Detect inconsistent data (marks should be between 0 and 100)
for cols in ['Maths','Science','English','Attendance']:
    print("\nInconsistent data in:", cols)
    print(df[(df[cols] < 0) | (df[cols] > 100)])

# Replace inconsistent values with NaN
for cols in ['Maths','Science','English','Attendance']:
    df.loc[(df[cols] < 0) | (df[cols] > 100), cols] = np.nan

# Check missing values again after cleaning
print("\nMissing values after replacing invalid data:")
print(df.isnull().sum())

# Fill missing values with mean
df['Maths'] = df['Maths'].fillna(df['Maths'].mean())
df['Science'] = df['Science'].fillna(df['Science'].mean())
df['English'] = df['English'].fillna(df['English'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())

print("\nDataset after handling missing values:")
print(df)
```

```python
# 5. Data Formatting and Normalization

# Check data types again
print("\nData types after cleaning:")
print(df.dtypes)

# Example: Ensure Attendance is float
df['Attendance'] = df['Attendance'].astype(float)

print("\nData types after conversion:")
print(df.dtypes)

# Normalization (Min-Max Scaling)
for cols in ['Maths','Science','English','Attendance']:
    df[cols] = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())

print("\nDataset after normalization:")
print(df)
```
